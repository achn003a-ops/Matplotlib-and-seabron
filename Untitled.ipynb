```python
# vizaulization library - 
# import matplotlib.pyplot as plt

# or - 
from matplotlib import pyplot as plt
```

```python
# line plot - 
study_hr = [2,4,6,8,9,10]
st1_score = [40,50,55,65,80,95]            # student 1
st2_score = [36,48,51,58,65,84]            # student 2
st3_score = [49,54,57,69,89,92]            # student 3

plt.figure(figsize=(5,4))
plt.plot(study_hr,st1_score,linestyle="--",color="red", marker="o", label="student1")                # plot is used for line plot
plt.plot(study_hr,st2_score,linestyle="-",color="blue", marker="*", label="student2") 
plt.plot(study_hr,st3_score,linestyle="-",color="green", marker="^", label="student3") 
plt.title("study_hr vs exam_score")       # name of the graph
plt.xlabel("study_hr")         # x_Axis name
plt.ylabel("exam_Score")         # y_axis name
plt.legend()                   # to show the labels in the graphs
plt.grid(True)
plt.show()                    
```

```python
# Scatter plot - 
study_hr = [2,4,6,8,9,10]
st1_score = [40,50,55,65,80,95]            # student 1
st2_score = [36,48,51,58,65,84]            # student 2

plt.figure(figsize=(5,4))
plt.scatter(study_hr,st1_score,marker="*",color="blue",s=100,edgecolors="black",alpha=0.1,label="student1")
plt.scatter(study_hr,st2_score,marker="o",color="red",s=100,edgecolors="black",label="student2")
plt.title("study_hr vs exam_score")       # name of the graph
plt.xlabel("study_hr")         # x_Axis name
plt.ylabel("exam_Score")         # y_axis name
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Bar graoh - used when we have one numerical and one categorial column

stores = ["A","B","C","D","E"]
sales1 = [56,89,23,74,65]
sales2=[59,83,30,80,75]

plt.figure(figsize=(5,6))
plt.bar(stores,sales1,color="blue",width=0.5,edgecolor="black")
plt.bar(stores,sales2,color="red",width=0.5,edgecolor="black")
plt.title("store_names vs sales")
plt.xlabel("Store_names")
plt.ylabel("Sales")

plt.grid(True)
plt.show()
```

```python
# grouped bar chart
import numpy as np
stores = ['a','b','c','d','e']

sales1 = [56,67,98,85,60]
sales2 = [23,60,76,67,35]

x=np.arange(len(stores))

width=0.4
# left bar
plt.bar(x-width/2,sales1,width,color='red',label='sale1')
# right bar
plt.bar(x+width/2,sales2,width,color='green',label='sale2')

plt.xticks(x,stores)
plt.title('stores_names vs sales')
plt.xlabel('stores_names')
plt.ylabel('sales')
plt.legend()
plt.show()
```

```python
# histogram
# when you want to check the frequency or values count in particular column

student=[2,1,5,6,3,4,8,5,6,9,8,7,4,5,6,3,2,1,7,8,8,2]
len(student)

plt.figure(figsize=(4,5))

plt.hist(student,bins=5,color="red")
plt.title("student frequency")
plt.xlabel("student")
plt.ylabel("frequency")
plt.show()
```
